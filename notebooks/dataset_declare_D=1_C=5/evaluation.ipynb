```python
import json
import pathlib

import pandas as pd

pd.set_option("max_colwidth", 400)

prefix = "dataset_declare_D=1_C=5"
separator = "_"
runs = [run for run in pathlib.Path("../../runs/").glob(f"{prefix}{separator}*")]
assert len(runs) == 5
for run in runs:
    print(run.name)
```

```python
def evaluate(run):
    print(
        f"Evaluating models trained on",
        prefix,
        run.stem.removeprefix(prefix + separator),
    )
    models = [args.parent for args in run.rglob("args.json")]

    results_records = []
    for model in models:
        args = json.loads(model.joinpath("args.json").read_text())
        # print(args["constraints"])
        # print(args.get("constraints_multiplier", None))
        results = model.glob("**/results.json").__next__()
        # print(model.name)
        # print(results.read_text())
        results = json.loads(results.read_text())
        results_records.append(
            {
                "model": model.name,
                "constraints": args["constraints"],
                "constraints_multiplier": args.get("constraints_multiplier", None),
                "acc": results["acc"],
                "dld": results["dld"],
                "norm_dld": results["norm_dld"],
                "weighted_norm_dld": results["weighted_norm_dld"],
            }
        )

    df = pd.DataFrame(results_records, index=[r["model"] for r in results_records])
    df = df.drop(columns=["model"])
    df = df.sort_values(by="dld", ascending=True)
    display(df)
```

```python
for run in runs:
    evaluate(run)
    print()
```
