```python
import json
import pathlib

import pandas as pd

base_model_path = pathlib.Path(
    "../../runs/helpdesk_1/100%/20250307.2050.no_constraint/"
)
constraints_task_fine_tuned_model_path = pathlib.Path(
    "../../runs/helpdesk_1/100%/20250308.1124.constraints/"
)
constraints_fine_tuned_model_path = pathlib.Path(
    "../../runs/helpdesk_1/100%/20250308.1247.constraints/"
)

alt_dataset_constraints_task_fine_tuned_model_path = pathlib.Path(
    "../../runs/helpdesk_2/100%/20250311.1316.constraints"
)
alt_dataset_constraints_fine_tuned_model_path = pathlib.Path(
    "../../runs/helpdesk_2/100%/20250311.1320.constraints"
)
alt_dataset_task_fine_tuned_model_path = pathlib.Path(
    "../../runs/helpdesk_2/100%/20250311.1316.no_constraint"
)
```

### Tests over `helpdesk_1`

```python
pd.DataFrame.from_records(
    [
        {"model": "base"}
        | json.loads(
            (
                base_model_path.rglob("*.best_val_acc/*/results.json")
                .__next__()
                .read_text()
            )
        ),
        {"model": "fine_tuned"}
        | json.loads(
            (
                constraints_task_fine_tuned_model_path.rglob(
                    "*.best_val_acc/*/results.json"
                )
                .__next__()
                .read_text()
            )
        ),
        {"model": "constraints_only_fine_tuned"}
        | json.loads(
            (
                constraints_fine_tuned_model_path.rglob("*.best_val_acc/*/results.json")
                .__next__()
                .read_text()
            )
        ),
        {"model": "alt_dataset_fine_tuned"}
        | json.loads(
            (
                alt_dataset_constraints_task_fine_tuned_model_path.rglob(
                    "*.best_val_acc/helpdesk_1/*/results.json"
                )
                .__next__()
                .read_text()
            )
        ),
        {"model": "constraints_only_alt_dataset_fine_tuned"}
        | json.loads(
            (
                alt_dataset_constraints_fine_tuned_model_path.rglob(
                    "*.best_val_acc/helpdesk_1/*/results.json"
                )
                .__next__()
                .read_text()
            )
        ),
        {"model": "task_only_alt_dataset_fine_tuned"}
        | json.loads(
            (
                alt_dataset_task_fine_tuned_model_path.rglob(
                    "*.best_val_acc/helpdesk_1/*/results.json"
                )
                .__next__()
                .read_text()
            )
        ),
    ],
    index="model",
)
```
